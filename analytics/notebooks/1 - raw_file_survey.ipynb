# Raw file survey

Read-only pass over every `.sas7bdat` file across all 5 raw PDS trial exports.
No analysis, no joins, no cleaning here — just: what columns exist, what does the data look like.
This is the input to the CDISC harmonization mapping in `dbt/models/staging/`.

```python
import pandas as pd
import pyreadstat
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

DATA_DIR = Path('..') / '..' / 'Data' / 'raw'
assert DATA_DIR.exists(), DATA_DIR
```

```python
def read_domain(trial_folder, filename):
    """Read one .sas7bdat file and print its shape + columns + head. Read-only, no transforms."""
    path = DATA_DIR / trial_folder / filename
    df, meta = pyreadstat.read_sas7bdat(str(path))
    print(f'--- {trial_folder}/{filename} ---')
    print('shape:', df.shape)
    print('columns:', list(df.columns))
    display(df.head())
    return df
```
