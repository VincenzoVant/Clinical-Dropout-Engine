# Column comparison across trials

Structural comparison only — which columns exist in which trial's files.
No row-level analysis. Built to make `raw_file_survey.ipynb`'s output diffable
instead of scrolling 71 tables by eye.

Two views:
1. **Presence matrix** for the 3 raw-domain trials (they share filenames: `demo`, `disposit`, `death`, `ae`, ...) — column x trial boolean table, sorted by how many trials share each column.
2. **Side-by-side column lists** for the 2 ADaM trials (`adsl`, `adae`) — these don't share a naming convention with the raw domains, so a boolean matrix isn't meaningful; just eyeball the two lists together.
